# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(code):
    position = code.index(':')
    supplier = code[:position]
    return supplier
```

```python
get_supplier_code('ACME:123-L')
```

```python
def get_product_num(code):
    start = code.index(':') + 1
    end = code.index('-')
    product = code[start:end]
    return product
```

```python
get_product_num('ACME:123-L')
```

```python
def get_size(code):
    position = code.index('-')
    size = code[position + 1:]
    return size
```

```python
get_size('ACME:123-L')
```

```python
part1 = 'ACME:123-L'
part2 = 'DI:12345-M'
part3 = 'ACE:1-12'
```

```python
print(f"For {part1}, the supplier is {get_supplier_code(part1)}. Part number {get_product_num(part1)} is available in size {get_size(part1)}.")
```

```python
print(f"For {part2}, the supplier is {get_supplier_code(part2)}. Part number {get_product_num(part2)} is available in size {get_size(part2)}.")
```

```python
print(f"For {part3}, the supplier is {get_supplier_code(part3)}. Part number {get_product_num(part3)} is available in size {get_size(part3)}.")
```

```python
part_codes = ['ACME:123-L', 'DI:12345-M', 'ACE:1-12']
```

```python
for code in part_codes:
    print(f"For {code}, the supplier is {get_supplier_code(code)}. Part number {get_product_num(code)} is available in size {get_size(code)}.")
```
