This lab focuses on defining custom functions in Python.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician or band is {music}."
```

```python
print(favorite_things("Your Name", "Your Favorite Movie", "Your Favorite Band"))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here("I joined Year Up United Data Analyst program to develop my data and analytical skills. To help me prepare for a career in data analytics."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("My house", ""))
```
