# Lab 2 - Exercise 2.B: Datetime Functions

This notebook practices using Python datetime functions.

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()

today
```

```python
print(today.strftime("Today is %A, %B %d, %Y"))
```

# My birthday

```python
birthday = datetime.date(2004, 12, 22)

birthday
```

```python
print(birthday.strftime("My birthday is %m/%d/%Y"))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)

ninety_d
```

```python
print(ninety_d.strftime("90 days from today is %B %d, %Y"))
```

# Time for Dinner

```python
dinner_time = datetime.time(7, 00)
dinner_time = datetime.time(19, 00)
```

```python
print(dinner_time.strftime("Let's meet for dinner at %I:%M %p"))
```
