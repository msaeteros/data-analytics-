# Exercise 1.A NumPy: Manipulating Arrays 

```python
import numpy as np
```

 ## Looping Through Arrays

```python
arrA = np.array([ 
    [11,12,13], 
    [14,15,16] 
])

print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop is used to go through the first dimension of the arraA array. While the seconf for loop through the second
dimension and prints each elements(number value) thats multiplied by 2. 

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This prints the entire row thats multiplied by 2 each time the inner loop runs
```

```python
## Joining and Stacking Arrays 
```

```python
joined_array_1 = np.concatenate((arrA, arrA), axis = 0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# This produces a ValueError because np.stack might require for both arrays to have the same elemensts(shape)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# This can produce a error because the arrays have different number of rows 
```

```python
horizontal_stack = np.hstack((arrA, arrA))
# hstack joins arrays horizontally.

vertical_stack = np.vstack((arrA, arrA))
# vstack joins arrays vertically.

depth_stack = np.dstack((arrA, arrA))
# dstack joins arrays third dimension

print(horizontal_stack)
print(vertical_stack)
print(depth_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)

# Contains 2 arrays. Each array has 2 dimensions.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# The result contains 3 arrays. Each array has 2 dimensions.
```
