# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np 
import pandas as pd
```

## Basic Aggregations

```python
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}
```

```python
df_sales = pd.DataFrame(data)
df_sales
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
total_units = df_sales['Units_Sold'].sum()
print('Total Units Sold:', total_units)
```

```python
total_revenue= df_sales['Revenue'].sum()
print('Total Revenue:', total_revenue)
```

```python
avg_units_sold = round(df_sales['Units_Sold'].mean(),2)
print('Average Units Sold:', avg_units_sold)
```

```python
avg_revenue = round(df_sales['Revenue'].mean(),2)

print('Average Revenue:', avg_revenue)
```

## Grouping Data

```python
product_totals = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()

print(product_totals)
```

```python
region_avg = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
print(region_avg)
```

```python
product_region =df_sales.groupby(['Product','Region'])[['Units_Sold','Revenue']].sum()
print(product_region)
```

## Sorting and Ranking Data

```python
df_sales_sorted = df_sales.sort_values(by ='Units_Sold',ascending = True)
print(df_sales_sorted)
```

```python
df_sales_sort = df_sales.sort_values(by ='Revenue',ascending = False)
print(df_sales_sort)
```

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False)

print(df_sales)
```

## Merging DataFrames

```python
product_data = {'Product': ['A', 'B', 'C'],
'Category': ['Electronics', 'Furniture', 'Appliances']}
df_products = pd.DataFrame(product_data)
```

```python
df_merged = pd.merge(df_sales, df_products, on='Product')

print(df_merged)
```

```python
category_revenue = df_merged.groupby('Category')['Revenue'].sum()

print(category_revenue)
```
