```python
import numpy as pd 
import matplotlib.pyplot as plt 
```

 ## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
```

```python
plt.bar(products, sales, color='skyblue', width=0.5)

plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')

plt.show()
```

```python
import numpy as np 
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.show()
```

```python
s = np.random.rand(50) * 100

plt.scatter(x, y, s=s, c=s, cmap='viridis')

plt.show()
```

## Subplots

```python
plt.figure(figsize=(10, 5))

# First subplot: Line chart
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('X axis')
plt.ylabel('Y axis')

data = np.random.randn(1000)

plt.subplot(1, 2, 2)
plt.hist(data)
plt.title('Histogram')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.tight_layout()

plt.show()
```
