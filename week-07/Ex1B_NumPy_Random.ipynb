## Generating Random Choices from an Array 

```python
students = np.array([
    'Jada', 'Idasha', 'Alvie', 'Christopher', 
    'Aalexus', 'Richard', 'Moses', 'Ardita', 
    'Arionsae', 'Dijeneba', 'Gbolahan', 'Denae'])
print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4), replace=False)
print(breakout_groups)

# This would be a effecitive way to creeate breakout groups because it does randomly assign people to a group 
# Dividing the groups evenly without repeating anyone
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.15, 0.10, 0.10, 0.10, 0.08,
       0.05, 0.04, 0.03, 0.02, 0.03, 0.00]
)

print(call_on_weighted)

# Probability :
# 
```

 ## Generating a Normal Distribution
