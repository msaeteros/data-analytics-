## Mathematical Operations

```python
print("r_1:", r_1)
print("Shape of r_1:", r_1.shape)

print("r_2:", r_2)
print("Shape of r_2:", r_2.shape)

result_1 = np.add(r_1, r_2)
print("Addition result:", result_1)
print("Shape of result:", result_1.shape)
```

```python
print("r_1:", r_1)
print("Shape of r_1:", r_1.shape)

print("r_3:", r_3)
print("Shape of r_3:", r_3.shape)

result_2 = np.add(r_1, r_3)
print("Addition result:", result_2)
print("Shape of result:", result_2.shape)
```

```python
print("r_2:", r_2)
print("Shape of r_2:", r_2.shape)

print("r_4:", r_4)
print("Shape of r_4:", r_4.shape)

result_3 = np.add(r_2, r_4)
print(result_3)

# When I run the code I get a ValueError because the shapes (1,2) and (4,5) since the columnd dont match  
```

```python
result_4 = np.add(np.array([1]), r_4)
print(result_4)
```

```python
result_5 = np.add(np.array([[1]]), r_4)
print(result_5)
```

```python
result_6 = np.add(np.array([1, 1, 1, 1, 1]), r_4)
print(result_6)
```

```python
result_7 = np.add(np.array([[1], [1]]), r_4)
print(result_7)

# This results in a ValueError showing that the code cant be broadcasted together since the rows dont match (2,1) and (4,5)
```

Broadcasting in NumPy allows combining two arrays together even when they have different shapes, as long
as the arrays share similar dimensions. In using the add function, Numpy adds the corresponding values from each array together.
The shape of the array is determined by how compatible the dimensions of two arrays are.

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])

bag2 = np.array([4, 0, 1, 2])

print("Bag 1:", bag1)
print("Bag 2:", bag2)
```

```python
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)

print(f"The first bag contains {total_bag1} items, and the second bag contains {total_bag2} items.")
```

```python
total_categories = np.add(bag1, bag2)

print(f"There are {total_categories[0]} vegetables, {total_categories[1]} fruits, {total_categories[2]} boxes of snacks, and {total_categories[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])

print(ali)
```

```python
ben = np.array([5, 5, 3, 4])

print(ben)
```

```python
carla = np.array([3, 6, 0, 3, 4])

print(carla)
```

```python
all_knocks = np.array([
    ali,
    [0, 5, 5, 3, 4],
    carla
])

print(all_knocks)
```

```python
ali_total = np.sum(all_knocks[0])
ben_total = np.sum(all_knocks[1])
carla_total = np.sum(all_knocks[2])

group_total = np.sum(all_knocks)

print("Ali knocked on", ali_total, "doors.")
print("Ben knocked on", ben_total, "doors.")
print("Carla knocked on", carla_total, "doors.")
print("The group collectively knocked on", group_total, "doors.")
```
